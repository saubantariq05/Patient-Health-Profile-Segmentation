## FEATURE SELECTIONS


```python
features = [
    'HighBP',
    'HighChol',
    'BMI',
    'Stroke',
    'HeartDiseaseorAttack',
    'GenHlth',
    'DiffWalk',
    'Smoker',
    'PhysActivity',
    'Fruits',
    'Veggies',
    'HvyAlcoholConsump',
    'MentHlth',
    'PhysHlth',
    'Age'
]
```

```python
X =df[features]
X.head()
```

```python
X.shape
```

```python
X.columns
```

```python
print("Diabetic012" in X.columns)
```

```python
X.describe().T
```

```python
from sklearn.preprocessing import MinMaxScaler
```

```python
scaler = MinMaxScaler()
```

```python
X_scaled = scaler.fit_transform(X)
```

```python
print(X_scaled.shape)
print(X_scaled.min(axis=0))
print(X_scaled.max(axis=0))
```

```python
from sklearn.cluster import KMeans
```

```python
wcss =[]

for k in range(2,11):
    kmeans =KMeans(n_clusters=k)
    kmeans.fit(X_scaled)
    wcss.append(kmeans.inertia_)
```

```python
plt.plot(range(2,11),wcss,marker='o')
plt.title('Elbow Method')
plt.xlabel('number of cluster')
plt.ylabel('wcss/inertia')
plt.show()
```

```python
from sklearn.metrics import silhouette_score
```

```python
rng  = np.random.RandomState(42)
sample_indices = rng.choice(
    X_scaled.shape[0],
    size =10000,
    replace =False
    
)

X_sample = X_scaled[sample_indices]
```

```python
for k in [5, 6, 7]:
    kmeans = KMeans(
        n_clusters=k,
        random_state=42,
        n_init=10
    )

    labels = kmeans.fit_predict(X_scaled)

    score = silhouette_score(
        X_sample,
        labels[sample_indices]
    )

    print(f"K = {k}, Silhouette Score = {score:.4f}")
```

```python
kmeans_5 = KMeans(n_clusters =5,random_state =42,n_init =10)
```

```python
cluster_labels = kmeans_5.fit_predict(X_scaled)
```

```python
df_profile = df.copy()
```

```python
df_profile['Cluster'] = cluster_labels
```

```python
from scipy.stats import chi2_contingency

chi2, p_value, dof, expected = chi2_contingency(
    diabetes_counts
)

print("Chi-square statistic:", chi2)
print("P-value:", p_value)
print("Degrees of freedom:", dof)
```

```python
import numpy as np

n = diabetes_counts.sum().sum()

cramers_v = np.sqrt(
    chi2 / (
        n * min(
            diabetes_counts.shape[0] - 1,
            diabetes_counts.shape[1] - 1
        )
    )
)

print("Total observations:", n)
print("Cramer's V:", round(cramers_v, 4))
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score

kmeans_A = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=10
)
labels_A = kmeans_A.fit_predict(X_scaled)

kmeans_B = KMeans(
    n_clusters=5,
    random_state=100,
    n_init=10
)
labels_B = kmeans_B.fit_predict(X_scaled)

ari = adjusted_rand_score(labels_A, labels_B)

print("Adjusted Rand Index:", round(ari, 4))
```

```python
from sklearn.cluster import KMeans
from sklearn.metrics import adjusted_rand_score

random_states = [100, 202, 7, 123]

for seed in random_states:
    model = KMeans(
        n_clusters=5,
        random_state=seed,
        n_init=10
    )

    labels_new = model.fit_predict(X_scaled)

    ari = adjusted_rand_score(labels_A, labels_new)

    print(f"Seed = {seed}, ARI = {ari:.4f}")
```

```python


# Create a reference model with more initializations
kmeans_ref = KMeans(
    n_clusters=5,
    random_state=42,
    n_init=20
)

labels_ref = kmeans_ref.fit_predict(X_scaled)

print("Reference inertia:", round(kmeans_ref.inertia_, 2))

# Compare other random seeds
for seed in [100, 202, 7, 123]:
    model = KMeans(
        n_clusters=5,
        random_state=seed,
        n_init=20
    )

    labels_new = model.fit_predict(X_scaled)

    ari = adjusted_rand_score(labels_ref, labels_new)

    print(
        f"Seed = {seed}, "
        f"ARI = {ari:.4f}, "
        f"Inertia = {model.inertia_:.2f}"
    )
```

```python
print(
    "ARI: seed 100 vs seed 7:",
    round(adjusted_rand_score(labels_ref, labels_ref), 4)
)
```

```python
kmeans_final = KMeans(
    n_clusters=5,
    random_state=100,
    n_init=20
)

cluster_labels_final = kmeans_final.fit_predict(X_scaled)

df_profile = df.copy()
df_profile['Cluster'] = cluster_labels_final

print("Final inertia:", round(kmeans_final.inertia_, 2))
print(df_profile['Cluster'].value_counts().sort_index())
```

```python
# Calculate the new cluster sizes
cluster_sizes_final = (
    df_profile['Cluster']
    .value_counts()
    .sort_index()
)

print("Cluster sizes:")
print(cluster_sizes_final)

# Calculate the new cluster profiles
cluster_profile_final = (
    df_profile.groupby('Cluster')[features]
    .mean()
    .round(2)
)

cluster_profile_final
```

```python
diabetes_counts_final = pd.crosstab(
    df_profile['Cluster'],
    df_profile['Diabetes_012']
)

diabetes_percentages_final = pd.crosstab(
    df_profile['Cluster'],
    df_profile['Diabetes_012'],
    normalize='index'
) * 100

diabetes_percentages_final.round(2)
```

```python
from scipy.stats import chi2_contingency

chi2_final, p_value_final, dof_final, expected_final = (
    chi2_contingency(diabetes_counts_final)
)

print("Chi-square statistic:", chi2_final)
print("P-value:", p_value_final)
print("Degrees of freedom:", dof_final)
```

```python
import numpy as np

n_final = diabetes_counts_final.sum().sum()

cramers_v_final = np.sqrt(
    chi2_final / (
        n_final * min(
            diabetes_counts_final.shape[0] - 1,
            diabetes_counts_final.shape[1] - 1
        )
    )
)

print("Total observations:", n_final)
print("Cramer's V:", round(cramers_v_final, 4))
```

```python
# Plot 1: Cluster sizes
plt.figure(figsize=(8, 5))

sns.countplot(
    data=df_profile,
    x='Cluster',
    order=sorted(df_profile['Cluster'].unique())
)

plt.title('Number of Respondents in Each Health Profile Cluster')
plt.xlabel('Cluster')
plt.ylabel('Number of Respondents')
plt.tight_layout()
plt.show()


# Plot 2: Diabetes category distribution
diabetes_percentages_final.plot(
    kind='bar',
    stacked=True,
    figsize=(9, 5)
)

plt.title('Diabetes Category Distribution Across Final Clusters')
plt.xlabel('Cluster')
plt.ylabel('Percentage of Respondents')
plt.xticks(rotation=0)

plt.legend(
    title='Diabetes Category',
    labels=[
        'No diabetes (0)',
        'Prediabetes (1)',
        'Diabetes (2)'
    ]
)

plt.tight_layout()
plt.show()
```
