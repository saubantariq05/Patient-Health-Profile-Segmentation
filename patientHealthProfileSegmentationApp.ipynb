```python

import joblib

# Load the saved objects into new variables
loaded_scaler = joblib.load("models/scaler.pkl")
loaded_kmeans = joblib.load("models/kmeans_model.pkl")

# Verify the loaded objects
print("Scaler features:", loaded_scaler.n_features_in_)
print("K-Means features:", loaded_kmeans.n_features_in_)
print("Number of clusters:", loaded_kmeans.n_clusters)
print("Model inertia:", round(loaded_kmeans.inertia_, 2))
```

```python

features = [
    'HighBP', 'HighChol', 'BMI', 'Stroke',
    'HeartDiseaseorAttack', 'GenHlth', 'DiffWalk',
    'Smoker', 'PhysActivity', 'Fruits', 'Veggies',
    'HvyAlcoholConsump', 'MentHlth', 'PhysHlth', 'Age'
]
```

```python

import pandas as pd



df = pd.read_csv("diabetes_012_health_indicators_BRFSS2015.csv")

print("Dataset shape:", df.shape)
print("Dataset loaded successfully!")
```

```python

sample_input = df[features].iloc[[0]]

sample_scaled = loaded_scaler.transform(sample_input)

sample_cluster = loaded_kmeans.predict(sample_scaled)

print("Input shape:", sample_input.shape)
print("Scaled input shape:", sample_scaled.shape)
print("Assigned cluster:", sample_cluster[0])
```

```python

def predict_cluster(input_data):
    # Convert the 15 values into ONE ROW with 15 columns
    input_df = pd.DataFrame([input_data], columns=features)

    # Scale using the saved scaler
    input_scaled = loaded_scaler.transform(input_df)

    # Predict the cluster
    cluster = loaded_kmeans.predict(input_scaled)

    return int(cluster[0])
```

```python

# Use the first row of our existing dataset
sample_values = df[features].iloc[0].tolist()

result = predict_cluster(sample_values)

print("Predicted cluster:", result)
```
